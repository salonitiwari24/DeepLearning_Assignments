```python
# import libraries

import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
# Load Fashion MNIST Dataset
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.fashion_mnist.load_data()
```

```python
# Normalize the Data
train_images = train_images / 255.0
test_images = test_images / 255.0
```

```python
# Display Dataset Information
print("Training Images Shape:", train_images.shape)
print("Testing Images Shape:", test_images.shape)

print("Training Labels Shape:", train_labels.shape)
print("Testing Labels Shape:", test_labels.shape)
```

```python
# Define Class Names
class_names = [
    "T-shirt/Top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle Boot"
]
```

```python
# Visualize Sample Images
plt.figure(figsize=(6,6))

for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(train_images[i], cmap="gray")
    plt.title(class_names[train_labels[i]])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Display One Image
plt.imshow(train_images[7], cmap="gray")
plt.title(class_names[train_labels[7]])
plt.axis("off")
plt.show()
```

```python
# Define Neural Network Model
my_model = tf.keras.models.Sequential()

my_model.add(tf.keras.layers.Flatten(input_shape=(28, 28)))
my_model.add(tf.keras.layers.Dense(128, activation="relu"))
my_model.add(tf.keras.layers.Dense(10, activation="softmax"))
```

```python
# Compile the Model
my_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Train the Model
history = my_model.fit(
    train_images,
    train_labels,
    epochs=3
)
```

```python
# Evaluate the Model
test_loss, test_accuracy = my_model.evaluate(test_images, test_labels)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```
